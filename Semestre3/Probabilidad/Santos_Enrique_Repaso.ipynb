```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
from scipy.special import gamma
import scipy 
from scipy import stats
 
```

```python
def gauss(x, mu, sigma):
    return 1.0/(np.sqrt(2.0*np.pi)*sigma)*np.exp(-(x - mu)**2/(2.0*sigma**2))
```

```python
def tdist(t,nu):
    return  gamma((nu+1)*0.5 )/(gamma(nu*0.5)*np.sqrt(np.pi*nu)) * (1.0 +t**2/nu) **(-(nu+1.0)*0.5)
```

```python
def fdist(f,nu1,nu2):
    return gamma((nu1+nu2)*0.5) * (nu1/nu2)**(nu1*0.5)/(gamma(nu1*0.5) * gamma(nu2*0.5)) *f**(nu1*0.5 -1.0)/(1.0+nu1*f/nu2) **((nu1+nu2) *0.5)
    
```

```python
def chi(x,nu):
    return 1.0/(2**(nu*0.5)*scipy.special.gamma(nu*0.5))*x**(nu*0.5 -1.0)*np.exp(-x*0.5)
```

```python

def t_dist_value(probabilidad, v1,v2):
    t_arr = np.linspace(0,1000,1000)
    ss = 0
    for tval in t_arr:
        prob = integrate.quad(fdist, 0.0, tval, args=(v1, v2))
        if prob[0] >= probabilidad:
            return t_arr[ss-1]
            break
        ss+=1
```

```python
valores = [
    (0.95, 1, 15),
    (0.95, 15, 7),
    (0.99, 24, 19),
    (0.05, 19, 24),
    (0.01, 28, 12)
]
for probabilidad, v1, v2 in valores:
    print(t_dist_value(probabilidad, v1, v2))
```

```python
n1, x1, s1 = 300, 102300, 5700  # Norte
n2, x2, s2 = 300, 98500, 3800  # Oeste
confianza = 0.99

diff_medias = x1 - x2

se = np.sqrt((s1**2 / n1) + (s2**2 / n2))

alpha = 1 - confianza
z_critico = stats.norm.ppf(1 - alpha / 2)

me = z_critico * se
lim_inf = diff_medias - me
lim_sup = diff_medias + me

print("=== EJERCICIO 14 (a): INTERVALO DE CONFIANZA DEL 99% ===")
print(f"Diferencia de Medias (x1 - x2): ${diff_medias:,.2f}")
print(f"Error Estándar (SE): ${se:,.4f}")
print(f"Valor Crítico Z (99%): {z_critico:.4f}")
print(f"Margen de Error (ME): ${me:,.2f}")
print(f"Intervalo de Confianza del 99%: [${lim_inf:,.2f}, ${lim_sup:,.2f}] dólares\n")

# 3. INCISO (b): Validación de supuestos poblacionales
print(f"Tamaño de muestra n1 = {n1} >= 30 y n2 = {n2} >= 30.\n")

# 4. INCISO (c): Validación del supuesto de varianzas
razon_varianzas = (s1**2) / (s2**2)
print("=== EJERCICIO 14 (c): SUPUESTO DE VARIANZAS ===")
print(f"Varianza Norte (s1^2): {s1**2:,}")
print(f"Varianza Oeste (s2^2): {s2**2:,}")
print(f"Razón de Varianzas (s1^2 / s2^2): {razon_varianzas:.2f}")
print("No es razonable asumir varianzas iguales ya que s1^2 es más del doble de s2^2.")
```
