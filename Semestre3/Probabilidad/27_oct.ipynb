```python
import numpy as np
import scipy.integrate as integrate
```

```python
def normal(z):
  return (1.0 / np.sqrt(2.0 * np.pi)) * np.exp(-(z**2) / 2.0)
```

```python
z_arr = np.linspace(-10,10,1000)
ss = 0
for zval in z_arr:
    prob = integrate.quad(normal, -np.inf, zval)
    if prob[0] >= 0.95:
        print(z_arr[ss-1], ss)
        break
    ss+=1
```
