```python
import numpy as np
import scipy.integrate as integrate
import matplotlib.pyplot as plt
import scipy 
from scipy.stats import chi2
```

```python
def chi(x,nu):
    return 1.0/(2**(nu*0.5)*scipy.special.gamma(nu*0.5))*x**(nu*0.5 -1.0)*np.exp(-x*0.5)
```

```python
n = 10
v = n - 1
x_data = np.asarray([46.4, 46.1, 45.8, 47.0, 46.1, 45.9, 45.8, 46.9, 45.2, 46.0])
data_mean = np.mean(x_data)
s2 = np.var(x_data, ddof=1)
sigma = 0.05
chi2_inf = chi2.ppf(sigma/2, v)
chi2_sup = chi2.ppf(1-sigma/2, v)

print(data_mean)
print( s2)
```

```python
nu=9
t_arr = np.linspace(0,10,1000)

ss = 0
for tval in t_arr:
    prob = integrate.quad(chi, 0, tval, args=(nu))
    if prob[0] >= 0.025:
        print(t_arr[ss-1])
        break
    ss+=1
```

```python
from scipy.special import gamma
```

```python
def fdist(f,nu1,nu2):
    return gamma((nu1+nu2)*0.5) * (nu1/nu2)**(nu1*0.5)/(gamma(nu1*0.5) * gamma(nu2*0.5)) *f**(nu1*0.5 -1.0)/(1.0+nu1*f/nu2) **((nu1+nu2) *0.5)
```

```python
n1 = 15
n2= 12
v1= n1-1
v2=n2-1

t_arr = np.linspace(0,50,1000)
ss = 0
for tval in t_arr:
    prob = integrate.quad(fdist, 0.0, tval, args=(v1, v2))
    if prob[0] >= 0.990:
        print(t_arr[ss-1])
        break
    ss+=1
```
