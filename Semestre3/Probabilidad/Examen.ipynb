```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
from scipy.special import gamma as gamma 
```

```python
def tdist(t,nu):
    return  gamma((nu+1)*0.5 )/(gamma(nu*0.5)*np.sqrt(np.pi*nu)) * (1.0 +t**2/nu) **(-(nu+1.0)*0.5)
```

```python
def gauss(x_arr, mu, sigma):
    return 1.0/(np.sqrt(2.0*np.pi)*sigma)*np.exp(-(x_arr - mu)**2/(2.0*sigma**2))
```

```python
def fdist(f,nu1,nu2):
    return gamma((nu1+nu2)*0.5) * (nu1/nu2)**(nu1*0.5)/(gamma(nu1*0.5) * gamma(nu2*0.5)) *f**(nu1*0.5 -1.0)/(1.0+nu1*f/nu2) **((nu1+nu2) *0.5)
    
```

```python
#Pregunta 1
n1 = 10
n2 = 8
v1 = n1 - 1   
v2 = n2 - 1  

 
t_arr = np.linspace(0, 10, 10000)

for tval in t_arr:
    prob = integrate.quad(fdist, 0, tval, args=(v2, v1))

    if prob[0] >= 0.01:
        print(tval)
        break
```
