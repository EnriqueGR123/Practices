### Parte 1: Importación y Exploración Inicial
Importa la librería pandas.
Carga un archivo llamado datos_desordenados.csv usando pd.read_csv().
Explora tu DataFrame mostrando las primeras 5 filas con head() y revisa los tipos de datos y valores nulos con info().

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('datos.csv')
df.head()
```

```python
print(f'Tamaño del datafreme: {df.shape} \n')
print(df.info())
```

### Parte 2: Limpieza de Datos
Valores vacíos: Usa dropna() para eliminar las filas irreparables, o fillna() para reemplazar valores vacíos-
Duplicados: Elimina las filas repetidas usando drop_duplicates().
Corrección de formato: Si alguna columna numérica está como texto, conviértela usando astype().

```python
df = df.dropna()#Elimina valores nulos 
df = df.drop_duplicates() # Eliminar duplicados
df['Date'] = pd.to_datetime(df['Date'], format='mixed')
print(f'Tamaño del datafreme: {df.shape} \n')
```

### Parte 3: Análisis y Estadística
Genera un resumen estadístico rápido de todas las columnas con describe().
Encuentra qué variables están más relacionadas entre sí utilizando la matriz de correlaciones con corr().

```python
print(f'Resumen estadistico:\n {df.describe()}')
print(f'Correlaciones:\n {df.corr()} ')
```

### Parte 4: Agrupación de Datos 
Agrupa los datos usando df.groupby(), buscando que las agrupaciones muestren información relevante.

```python
df.groupby('Duration')['Calories'].sum()
```

### Parte 5: Visualización

Crea un gráfico de dispersión (scatter) para ver la relación entre la duración del ejercicio y las calorías quemadas.
